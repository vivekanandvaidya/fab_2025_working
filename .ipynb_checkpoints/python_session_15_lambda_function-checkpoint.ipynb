**lambda function**

- lambda function represent function concept 
 - we can right in single line 
- like list comprehension lambda function similar only 
- it will decrease time complexcity
 - always remember if you use many for loops  or many  condition using multiple line the timw complexciry will increase

**it will use a keyword lambda**

```python
# lambda <argument> : <expression>
```

```python
def summ(n):
    return(n+10)

summ(10)
```

$pattern-1$

**function with only one argument**

 - function name 
 - argument name 
 - return output

```python
# syntax : <function name> = lambda <argument neme>: <return output>

def summ(n):
    return(n+10)
summ(10)

# function name = summ
# argument name = n
# return output = n+10
```

```python
summ=lambda n:n+10
summ(100)
```

```python
def cube(n):
    return(n**3)
    cube(10)
```

```python
cube=lambda c:c**3
cube(2)
```

**pattern-2**

**two arguments**

```python
# syntax : <function name>= lambda <arg1>,<arg2>: <retuurn oitput>

def add(a,b):
    return(a+b)
add(50,50)

# function name = add
# arg1 a
# arg2 b
# return a+b
```

```python
add=lambda n1,n2: n2+n2
add(60,40)
```

```python
arg=lambda a,b,c : round((a+b+c)/3,2)
arg(40,50,60)
```

$pattern-3$

**default argumenr**

```python
avg=lambda a,b,c=500:round((a+b+c)/3,2)
avg(40,50,60)
```

$pattern-4$

**if-else**

```python
def max(a,b):
    if a>b:
        return(a)
    else:
        return(b)

max(10,20)
```

```python
def min(a,b):
    if a<b:
        return(a)
    else:
        return(b)

min(10,20)
```

```python
# syntax: function name = lambda <arg1>,<arg2>: <list comprehension>
# syntax = function name = lambda <arg1> <arg2> : <if_uot> <if_con><else><else_out>

maxx=lambda a,b:a if a>b else b
max(55,65)
```

**reduce**

 - all inbuild function can achive by reduce
 - reduce is available from function tools package
 - level 1 : reduc(lambda summ,i:summ+i,l1)
 - level 2 : reduce (lambda summ, i:summ+i,l1,initial_value)
 - for ex. we want initialize summ=0 then we choose level-1
 - for ex. we want start with other than zero then we choos level-2

```python
import functools
```

```python
dir (functools)
```

```python
import functools
l1=[1,2,3,4,5]
functools.reduce(lambda summ,i:summ+i,l1)
```

```python
from functools import reduce
l1=[1,2,3,4,5]
reduce(lambda summ,i:summ+i,l1)
```

```python
import functools as ft
l1=[1,2,3,4,5]
ft.reduce(lambda summ,i:summ+i,l1)
```
